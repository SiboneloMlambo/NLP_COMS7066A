Use a CPU runtime and run the cells in order. Full run: about 65 minutes; keep the tab open.

```python
# 1. Get the code (clones the repo on first run) and move into the project folder
import os
REPO_URL = 'https://github.com/SiboneloMlambo/NLP_COMS7066A.git'
REPO_DIR = '/content/NLP_COMS7066A'
if not os.path.isdir(REPO_DIR):
    !git clone -q {REPO_URL} {REPO_DIR}
else:
    !git -C {REPO_DIR} pull -q
%cd /content/NLP_COMS7066A/Project2_isiXhosa
assert os.path.exists('reproduction.sh'), 'reproduction.sh not found - check the clone above'
```

```python
# 2. Optional: use course zips stored in Google Drive (MyDrive)
USE_DRIVE = False
if USE_DRIVE:
    from google.colab import drive
    drive.mount('/content/drive')

# Check the course zips can be found before starting the long run
import glob
zips = [z for d in ['/content', '/content/drive/MyDrive'] for z in glob.glob(f'{d}/*.zip')]
have_txt  = any('TxtFiles1870-1894' in z and 'TxtLang' not in z for z in zips)
have_lang = any('TxtLangFiles1870-1894' in z for z in zips)
print('Zips found:', *zips, sep='\n  ')
assert have_txt and have_lang, ('Upload 1._TxtFiles1870-1894_TB.zip and TxtLangFiles1870-1894_TB.zip '
                                'to /content (or MyDrive with USE_DRIVE = True), then re-run this cell.')
```

```python
# 3. Run. MODE: '--smoke' (~10 min code check), '--fast' (~35 min, no fastText), or '' (full, ~65 min)
MODE = ''
import subprocess
cmd = ['bash', 'reproduction.sh', *MODE.split(), '--zips', '/content']
proc = subprocess.Popen(cmd, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True)
for line in proc.stdout:
    print(line, end='')
if proc.wait() != 0:
    raise RuntimeError('reproduction.sh failed - see the output above')
```

```python
# 4. Main results table and figures
from IPython.display import Image, display
out = 'outputs_smoke' if 'smoke' in MODE else 'outputs'
print(open(f'{out}/table_main.tex').read())
for f in ['fig1_ladder', 'fig2_eras', 'figS1_noise', 'figS2_surface']:
    display(Image(f'{out}/figures/{f}.png'))
```
